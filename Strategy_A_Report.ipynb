```python
# Strategy A - final report
#   pass@1                     = solved tasks / total tasks
#   Execution time (System A)  = Et of task1 + Et of task2 + ... + Et of taskN
#   Avg execution time         = Execution time (System A) / total number of tasks
#   Generation time (System A) = time the model spent writing all answers
import csv
import glob
import os

RESULTS = "results"            # folder with the Strategy A CSV files
PATTERN = "*initial_10.csv"    # change to "*final_30.csv" for the 30-task runs

for path in sorted(glob.glob(os.path.join(RESULTS, "oneshot_" + PATTERN))):
    rows = list(csv.DictReader(open(path, encoding="utf-8")))
    n = len(rows)
    solved = sum(int(r["solved"]) for r in rows)
    exec_times = [float(r["exec_time_s"]) for r in rows]
    total_exec = sum(exec_times)
    total_gen = sum(float(r["gen_time_s"]) for r in rows)

    print("=" * 60)
    print("Model:", rows[0]["model"] if rows else "?")
    print("File :", os.path.basename(path))
    print("-" * 60)
    print(f"{'Task':<12}{'Result':<16}{'Execution time (s)':>20}")
    for r, et in zip(rows, exec_times):
        print(f"{r['task_id']:<12}{r['status']:<16}{et:>20.3f}")
    print("-" * 60)
    print(f"Pass@1: {solved}/{n} = {solved / n:.2f}")
    print(f"Execution time (System A): {' + '.join(f'{et:.3f}' for et in exec_times)} = {total_exec:.3f} s")
    print(f"Avg execution time: {total_exec:.3f} / {n} = {total_exec / n:.3f} s")
    print(f"Generation time (System A): {total_gen:.3f} s (model writing all {n} answers)")
```
